```python
import pandas as pd #importing dataset
attributes = ["buying", "maint", "doorNum", "persons", "sizeBoot", "safety", "classValue"] #creating a list/name for each attribute
CarData = pd.read_csv("car.data", names=attributes) #reads the data from the data file into a data frame (table)
CarData["carID"]=["Car "+str(i) for i in range(1,len(CarData)+1)] #creates a unique ID for each car
```

```python
print ("Enter price of the car as low, med, high or vhigh: ") #price prompt for the user input
buying = input("buying: ") #variable to store user input
print ("User has input: ", buying)

print ("Enter the maintenance of the car as low, med, high or vhigh: ") #maint prompt for the user input
maint = input("maint: ") #variable to store user input
print ("User has input: ", maint)

print ("Enter safety rating of the car as low, med or high: ") #safety prompt for the user input
safety = input("safety: ") #variable to store user input
print ("User has input: ", safety)
```

```python
buying_scores = { #assigning numerical/integer values to each word
    "vhigh": 1, #high price scores low because it's not the most desirable
    "high": 2,
    "med": 3,
    "low": 4 #low price scores high because it would be preferable
}

maint_scores = { #assigning numerical/integer values to each word
    "vhigh": 1, #high maintenance scores low because the car's upkeep will cost a lot, so its less desirable
    "high": 2,
    "med": 3,
    "low": 4 #low maintenance scores high because its less hassle to run, so more desirable
}

safety_scores = { #assigning numerical/integer values to each word
    "low": 1, #safety scores are the opposite, low scores low because it's dangerous and undesirable
    "med": 2,
    "high": 3 #high safety scores highest because it is the most preferable option
}
```

```python
def calculate_score(buying, maint, safety): #creating a function to calculate the score
    buying_score = buying_scores[buying] #score given to it based on buying rating
    maint_score = maint_scores[maint] #score given to it based on maintenance rating (possibly dodgy)
    safety_score = safety_scores[safety] #score given to it based on safety rating

    total_score = buying_score + maint_score + safety_score #adds scores up, gives a total

    return total_score #returns the total

def score_classification(score): #another function to help link total & class
    if score >= 10:
        return "vgood" #if a score is greater than or equal to 10, its very good
    elif score >= 8:
        return "good" #if a score is greater than equal to 8, its good
    elif score >= 6:
        return "acc" #if a score is greater than or equal to 6, its acceptable
    else:
        return "unacc" #if its anything other than those values (so 5 or under) its unacceptable
```

```python
score = calculate_score(buying, maint, safety) #calls the calculate_score function
score_result = score_classification(score) #stores the total 

print("\nUSER INPUT RESULTS") #new line for neatness
print("Buying Score:", buying_scores[buying]) #shows each initial score, in this case buying
print("Maintenance Score:", maint_scores[maint]) #shows each initial score, in this case buying
print("Safety Score:", safety_scores[safety]) #shows each initial score, in this case buying
print("Total Score:", score, "/ 11") #shows the total score out of the max possible score
print("Classification:", score_result) #gives the predicted classification of the car
```

```python
import random
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

random.seed(38)  #makes random shuffle create the same result every run
shuffled = CarData[attributes].values.tolist()  #puts data from dataset into new list. [attributes] makes it only use things listed in the 'attributes'
random.shuffle(shuffled)  #shuffles data in said list
cut = int(0.7 * len(shuffled))  #70% training data and 30% testing data
train, test = shuffled[:cut], shuffled[cut:]  #splits the data into train or test

print("\nTraining records:", len(train)) #prints how many cars were used for training
print("Testing records:", len(test)) #prints how many cars were used for testing

actual = []  #stores all of the actual classifications of each car
predictions = []  #stores what the numerical scoring system predicts

for row in test:
    score = calculate_score(row[0], row[1], row[5])
    score_result = score_classification(score)
    actual.append(row[6]) #brings the actual classifications into the 'actual' row from earlier, for comparison
    predictions.append(score_result) #stores the score result

#accuracy stuff
accuracy = accuracy_score(actual, predictions) #using scikit, compares the two lists and calculates how many were correct (so i dont have to do maths)

print("\nTEST RESULTS") #\n for new line, for neatness
print("Correct Predictions:", sum(a == p for a, p in zip(actual, predictions))) #zip pairs the actual & prediction, a==p checks if actual & prediction match, sum counts the correct ones
print("Total Predictions:", len(actual)) #counts how many cars are in 'actual'
print("Accuracy:", round(accuracy * 100, 2), "%") #shows the accuracy as a percentage


```
